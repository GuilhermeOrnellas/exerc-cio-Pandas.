```python
import pandas as pd
import numpy as np


np.random.seed(42)
dates = pd.to_datetime(pd.date_range(start='2022-01-01', periods=100, freq='D'))
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']

df = pd.DataFrame({
    'data': np.random.choice(dates, 500),
    'valor_venda': np.random.randint(10, 500, 500) + np.random.rand(500),
    'categoria': np.random.choice(categorias, 500),
    'valor': np.random.randint(5, 200, 500) + np.random.rand(500)
})

df['data'] = pd.to_datetime(df['data'])

display(df.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
num_entries = 12 * 3
dates = pd.date_range(start='2021-01-01', periods=num_entries, freq='M')

df = pd.DataFrame({
    'data': dates,
    'valor_venda': np.random.randint(1000, 5000, size=num_entries) * np.random.rand(num_entries) + 5000,
    'quantidade': np.random.randint(50, 300, size=num_entries)
})

df['valor_venda'] = df['valor_venda'] * (1 + np.sin(df.index / 3) * 0.5)
df['quantidade'] = df['quantidade'] * (1 + np.cos(df.index / 2) * 0.3)

df['valor_venda'] = df['valor_venda'].round(2)
df['quantidade'] = df['quantidade'].astype(int)

display(df.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
num_entries = 12 * 3
dates = pd.date_range(start='2021-01-01', periods=num_entries, freq='M')

categories = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
states = ['SP', 'RJ', 'MG', 'BA']

df = pd.DataFrame({
    'data': dates,
    'valor_venda': np.random.randint(1000, 5000, size=num_entries) * np.random.rand(num_entries) + 5000,
    'quantidade': np.random.randint(50, 300, size=num_entries),
    'categoria': np.random.choice(categories, size=num_entries),
    'estado': np.random.choice(states, size=num_entries)
})

df['valor_venda'] = df['valor_venda'] * (1 + np.sin(df.index / 3) * 0.5)
df['quantidade'] = df['quantidade'] * (1 + np.cos(df.index / 2) * 0.3)

df['valor_venda'] = df['valor_venda'].round(2)
df['quantidade'] = df['quantidade'].astype(int)

display(df.head())
```
