```python
import pandas as pd

df = pd.read_csv('/content/qualidade_ar_brasil.csv')

print("Dimensões (linhas, colunas):", df.shape)


display(df.head())


df.info()


display(df.describe())


print("Valores ausentes por coluna:\n", df.isnull().sum())

print("Cidades únicas:", df['cidade'].unique())

medias_poluentes = df.groupby('cidade')[['pm25', 'pm10', 'no2', 'o3']].mean()
display(medias_poluentes)

iqa_cidades = (
    df.groupby('cidade')['indice_qualidade_ar']
    .mean()
    .sort_values(ascending=False)
)
display(iqa_cidades)

df_alerta = df[df['nivel_alerta'].isin(['Ruim', 'Muito Ruim'])]
display(df_alerta.head())


print("Registros críticos por cidade:\n", df_alerta['cidade'].value_counts())
print("Total de alertas críticos:", len(df_alerta))

piores_dias = df.sort_values('indice_qualidade_ar', ascending=False).head(10)
display(
    piores_dias[
        ['data', 'cidade', 'indice_qualidade_ar', 'pm25', 'pm10', 'no2', 'o3']
    ]
)

df['data'] = pd.to_datetime(df['data'])
df['mes'] = df['data'].dt.month


iqa_mensal = df.groupby('mes')['indice_qualidade_ar'].mean()
display(iqa_mensal)

matriz_corr = df[['temperatura', 'pm25', 'pm10', 'no2', 'o3']].corr()
display(matriz_corr)
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)
datas = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
cidades_uf = [
    ('São Paulo', 'SP'),
    ('Rio de Janeiro', 'RJ'),
    ('Belo Horizonte', 'MG'),
    ('Curitiba', 'PR'),
    ('Manaus', 'AM'),
]

dados = []
for d in datas:
  for cidade, uf in cidades_uf:
    sazonalidade = 1.4 if d.month in [6, 7, 8, 9] else 1.0
    fator_cidade = (
        1.3 if cidade == 'São Paulo' else (0.8 if cidade == 'Manaus' else 1.0)
    )

    pm25 = np.random.uniform(10, 45) * sazonalidade * fator_cidade
    pm10 = pm25 * np.random.uniform(1.5, 2.2)
    no2 = np.random.uniform(15, 60) * fator_cidade
    o3 = np.random.uniform(20, 70) / (sazonalidade**0.5)
    temp = np.random.uniform(15, 32)
    umid = np.random.uniform(40, 90)

    iqa = (pm25 * 1.5 + pm10 * 0.8 + no2 * 0.7 + o3 * 0.5) / 3

    if iqa < 40:
      alerta = 'Bom'
    elif iqa < 70:
      alerta = 'Moderado'
    elif iqa < 95:
      alerta = 'Ruim'
    else:
      alerta = 'Muito Ruim'

    dados.append([
        d.strftime('%Y-%m-%d'),
        cidade,
        uf,
        round(pm25, 2),
        round(pm10, 2),
        round(no2, 2),
        round(o3, 2),
        round(temp, 1),
        round(umid, 1),
        round(iqa, 1),
        alerta,
    ])

df_gerado = pd.DataFrame(
    dados,
    columns=[
        'data',
        'cidade',
        'estado',
        'pm25',
        'pm10',
        'no2',
        'o3',
        'temperatura',
        'umidade',
        'indice_qualidade_ar',
        'nivel_alerta',
    ],
)
df_gerado.to_csv('qualidade_ar_brasil.csv', index=False)
print("Arquivo 'qualidade_ar_brasil.csv' gerado com sucesso!")
```
