```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

meus_filtros = {
    'userId': 2
}

meus_headers = {
    'User-Agent': 'MeuProjetoDev/1.0 (exercicio-python)'
}

resposta = requests.get(url, params=meus_filtros, headers=meus_headers, timeout=10)

if resposta.status_code == 200:
  dados = resposta.json()
  print("--- Requisição realizada com sucesso! ---")
  print(f"Status Code: {resposta.status_code}")
  print(f"Total de posts retornados com filtro: {len(dados)}")
  print("\nPrimeiro post da lista filtrada:")
else:
    print(f"Erro na requisição! Código de status: {resposta.status_code}")
```

```python
import requests
import pandas as pd


def requisicao_segura(url, modo="json"):
    try:
        resposta = requests.get(url, timeout=10)
        resposta.raise_for_status()

        if modo == "json":
            return resposta.json()
        elif modo == "binario":
            return resposta.content
    except requests.exceptions.RequestException:
        return None


ceps = ["01001000", "20040002", "70040900"]
lista_enderecos = []

for c in ceps:
    url_viacep = f"https://viacep.com.br/ws/{c}/json/"
    dados_cep = requisicao_segura(url_viacep, modo="json")

    if dados_cep and "erro" not in dados_cep:
        lista_enderecos.append(dados_cep)

df_enderecos = pd.DataFrame(lista_enderecos)
print(df_enderecos[["cep", "logradouro", "bairro", "localidade", "uf"]])

url_invalida = "https://viacep.com.br/ws/00000000000/json/"
requisicao_segura(url_invalida, modo="json")

url_imagem = "https://picsum.photos/400/400"
conteudo_imagem = requisicao_segura(url_imagem, modo="binario")

if conteudo_imagem:
    with open("imagem_aleatoria.jpg", "wb") as arquivo:
        arquivo.write(conteudo_imagem)
```

```python
import io
import requests
from bs4 import BeautifulSoup
import pandas as pd

url_robots = "https://books.toscrape.com/robots.txt"
resposta_robots = requests.get(url_robots, timeout=10)
print(resposta_robots.text)

url_books = "https://books.toscrape.com/"
resposta_books = requests.get(url_books, timeout=10)
soup = BeautifulSoup(resposta_books.text, "html.parser")

livros = []
elementos_livros = soup.select(".product_pod")[:5]

for elemento in elementos_livros:
    titulo = elemento.h3.a["title"]
    preco = elemento.select_one(".price_color").text
    livros.append({"Titulo": titulo, "Preco": preco})

df_livros = pd.DataFrame(livros)
df_livros.to_csv("livros.csv", index=False)

url_wiki = "https://pt.wikipedia.org/wiki/Lista_de_pa%C3%ADses_por_popula%C3%A7%C3%A3o"
headers = {"User-Agent": "MeuProjetoScraping/1.0"}
resposta_wiki = requests.get(url_wiki, headers=headers, timeout=10)

tabelas = pd.read_html(io.StringIO(resposta_wiki.text))
df_wiki = tabelas[0]
print(df_wiki.head())
```
